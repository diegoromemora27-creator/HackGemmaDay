## 3. Helpers núcleo

`gemma_generate` (texto), `gemma_multimodal_generate` (texto + imágenes), extracción robusta de JSON y transcripción de audio.

````python
def gemma_generate(system_prompt: str, user_prompt: str, max_new_tokens: int = 2600, temperature: float = 0.35) -> str:
    """Genera texto con Gemma 4.

    CAMBIO DE ESTA VERSION: antes el system prompt y el user prompt se
    concatenaban a mano dentro de un unico mensaje 'user' ('SYSTEM:\n...\n\n
    USER:\n...'). Eso funcionaba, pero Gemma 4 (a diferencia de Gemma 2/3)
    trae soporte NATIVO para el rol 'system', y el formato de tool calling
    nativo inserta la declaracion de herramientas justo dentro del turno
    'system'. Separar los roles aqui es lo que permite despues conectar
    tools reales sin tener que reescribir esta funcion.
    """
    messages = [
        {'role': 'system', 'content': system_prompt},
        {'role': 'user', 'content': user_prompt},
    ]
    inputs = processor.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors='pt'
    ).to(model.device)
    input_len = inputs['input_ids'].shape[-1]
    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=0.9,
        )
    return processor.decode(output_ids[0][input_len:], skip_special_tokens=True).strip()


def gemma_multimodal_generate(system_prompt: str, user_text: str, image_paths: List[str], max_new_tokens: int = 1800, temperature: float = 0.3) -> str:
    """Igual que gemma_generate pero con imagenes. Mismo cambio de roles: el
    system prompt ahora va en su propio mensaje 'role: system' en vez de
    quedar mezclado con el texto del usuario. El dtype se sigue pasando en
    .to(...) porque aqui si hay pixel_values (las imagenes) que necesitan
    coincidir con el dtype del modelo; en gemma_generate (solo texto) ese
    dtype no aplicaba a nada y se quito para simplificar.
    """
    content = []
    for path in image_paths:
        content.append({'type': 'image', 'image': str(path)})
    content.append({'type': 'text', 'text': user_text})
    messages = [
        {'role': 'system', 'content': system_prompt},
        {'role': 'user', 'content': content},
    ]
    inputs = processor.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors='pt'
    ).to(model.device, dtype=model.dtype if hasattr(model, 'dtype') else torch.bfloat16)
    input_len = inputs['input_ids'].shape[-1]
    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=0.9,
        )
    return processor.decode(output_ids[0][input_len:], skip_special_tokens=True).strip()


def extract_json_block(raw_text: str, required_keys: Optional[List[str]] = None, debug_label: str = '') -> dict:
    """Extrae un bloque JSON de la respuesta de Gemma de forma tolerante a texto
    explicativo antes/despues, bloques ```json ... ``` en cualquier posicion (no
    solo al inicio) y JSON casi valido (comas colgantes, comillas simples, etc.
    via json_repair).

    A proposito NO devuelve un dict vacio si falla: lanza ValueError con la
    salida cruda de Gemma incluida en el mensaje. Si devolvieramos {} aqui, Pydantic
    rellenaria cada campo con su valor por defecto (ej. nombre='Estudiante') sin
    ningun error visible, que es exactamente el bug de "no extrae nada" que
    reportaste: fallaba en silencio y el default lo disimulaba.
    """
    original = raw_text
    text = raw_text.strip()
    text = re.sub(r'```json', '', text, flags=re.IGNORECASE)
    text = text.replace('```', '')
    start = text.find('{')
    end = text.rfind('}')
    if start == -1 or end == -1 or end <= start:
        print(f"[extract_json_block:{debug_label}] Gemma no devolvio un bloque {{...}} reconocible. Salida cruda:\n{original[:800]}")
        raise ValueError(f"Gemma no devolvio JSON reconocible ({debug_label}). Salida cruda: {original[:300]}")

    candidate = text[start:end + 1]
    try:
        data = json.loads(candidate)
    except Exception:
        try:
            data = json.loads(repair_json(candidate))
        except Exception as e:
            print(f"[extract_json_block:{debug_label}] JSON irreparable: {e}\nCandidato:\n{candidate[:800]}")
            raise ValueError(f"JSON irreparable en la respuesta de Gemma ({debug_label}): {e}")

    if not isinstance(data, dict):
        raise ValueError(f"Se esperaba un objeto JSON y se obtuvo {type(data)} ({debug_label}).")

    if required_keys:
        faltantes = [k for k in required_keys if not str(data.get(k) or '').strip() and not data.get(k)]
        if faltantes:
            print(f"[extract_json_block:{debug_label}] Gemma dejo vacios/omitio estos campos clave: {faltantes}. "
                  f"JSON recibido: {json.dumps(data, ensure_ascii=False)[:800]}")

    return data


def transcribe_audio_safe(audio_path: Optional[str]) -> str:
    """Transcribe audio a texto usando Whisper."""
    if not audio_path:
        return ''
    try:
        result = asr(audio_path)
        return result['text'].strip()
    except Exception as e:
        return f'[error_transcribiendo_audio] {e}'
````

## 3 bis. Preparación para tool calling real + nota sobre audio nativo

Cambios de esta versión (preparación para conectar Notion con tool calling nativo de Gemma 4, sin implementar todavía ninguna tool concreta):

1. **Roles nativos (arriba, celda 9):** `gemma_generate` y `gemma_multimodal_generate` ahora mandan el system prompt y el user prompt como dos mensajes separados (`role: 'system'` / `role: 'user'`) en vez de concatenarlos a mano dentro de un único mensaje `user`. Gemma 4 sí soporta el rol de sistema de forma nativa (Gemma 2/3 no lo tenían, por eso existía el hack), y el formato de tool calling nativo inserta la declaración de herramientas justo dentro del turno `system` — así que esto es un prerequisito real, no solo prolijidad.
2. **`gemma_chat_with_tools` (celda de abajo):** función genérica que implementa el ciclo oficial de 3 turnos de Gemma 4 para tool calling (Model's Turn → Developer's Turn → Final Response). Todavía **no tiene ninguna tool conectada** — recibe `tools` (lista de funciones Python con docstring) y `tool_functions` (diccionario nombre → función real) como parámetros, así que cuando definamos las funciones de Notion en el siguiente paso, solo hay que pasárselas aquí; esta función no se vuelve a tocar.
3. **`gemma_audio_generate_experimental` (celda de abajo):** no está conectada a ningún endpoint todavía. Gemma 4 acepta audio de forma nativa (además de texto e imagen), así que en teoría podrías mandarle el audio directo aquí en vez de transcribirlo primero con Whisper (`transcribe_audio_safe`). Se deja documentada como opción para el futuro; tu pipeline actual (Whisper → texto → Gemma) sigue siendo el que usan todos los endpoints y no se modificó.


````python
def _extract_tool_calls_fallback(raw_text: str) -> List[dict]:
    """Red de seguridad si tu version de transformers todavia no trae
    processor.parse_response(...) para Gemma 4. Implementa el mismo regex que
    usa la documentacion oficial de Gemma 4 para leer bloques
    <|tool_call>call:nombre{args}<tool_call|> de la salida cruda del modelo.
    """
    def _cast(v):
        v = v.strip()
        try:
            return int(v)
        except ValueError:
            pass
        try:
            return float(v)
        except ValueError:
            pass
        return {'true': True, 'false': False}.get(v.lower(), v.strip("'\""))

    calls = []
    for name, args in re.findall(r"<\|tool_call>call:(\w+)\{(.*?)\}<tool_call\|>", raw_text, re.DOTALL):
        arguments = {
            k: _cast(v1 or v2)
            for k, v1, v2 in re.findall(r'(\w+):(?:<\|"\|>(.*?)<\|"\|>|([^,}]*))', args)
        }
        calls.append({'name': name, 'arguments': arguments})
    return calls


def gemma_chat_with_tools(
    system_prompt: str,
    user_prompt: str,
    tools: Optional[List[Any]] = None,
    tool_functions: Optional[Dict[str, Any]] = None,
    max_new_tokens: int = 800,
    max_new_tokens_final: int = 900,
) -> dict:
    """Ciclo generico de tool calling nativo de Gemma 4 (Model's Turn ->
    Developer's Turn -> Final Response, segun la documentacion oficial).

    NO es especifico de Notion ni de ninguna otra tool:
    - 'tools': lista de funciones Python (con docstring estilo Google) que
      Gemma puede decidir usar. Se le pasan tal cual a apply_chat_template;
      Gemma genera el JSON schema solo, leyendo la firma y el docstring.
    - 'tool_functions': diccionario {nombre_de_la_tool: funcion_real} que
      este helper ejecuta si Gemma pide usarla. Si Gemma pide una tool que no
      esta en este diccionario (p. ej. porque el perfil activo no la tiene
      permitida), se le informa el error a Gemma en vez de ejecutar nada.
    - Si 'tools' es None, se comporta igual que un gemma_generate normal (sin
      capacidad de tool calling): esto sirve para perfiles como 'nino' o
      'adulto_mayor', a los que no les vamos a dar acceso a Notion.

    Devuelve:
      {
        'respuesta': texto final para mostrarle/leerle al estudiante,
        'tool_calls': [{'name':..., 'arguments':...}, ...]  (vacio si no se uso ninguna tool),
        'tool_resultados': [{'name':..., 'response':...}, ...],
      }

    Este helper todavia no se llama desde ningun endpoint: falta definir las
    funciones de Notion (siguiente paso) y un endpoint /assistant que las use,
    siguiendo el mismo patron async con job_id que ya usas en /experience
    (para no bloquear el tunel de ngrok mientras Gemma genera).
    """
    messages = [
        {'role': 'system', 'content': system_prompt},
        {'role': 'user', 'content': user_prompt},
    ]

    def _generate(msgs, tok_limit, skip_special: bool):
        text = processor.apply_chat_template(
            msgs, tools=tools, tokenize=False, add_generation_prompt=True
        )
        inputs = processor(text=text, return_tensors='pt').to(model.device)
        input_len = inputs['input_ids'].shape[-1]
        with torch.no_grad():
            out = model.generate(
                **inputs,
                max_new_tokens=tok_limit,
                do_sample=True,
                temperature=0.3,
                top_p=0.9,
            )
        return processor.decode(out[0][input_len:], skip_special_tokens=skip_special)

    if not tools:
        respuesta = _generate(messages, max_new_tokens, skip_special=True)
        return {'respuesta': respuesta.strip(), 'tool_calls': [], 'tool_resultados': []}

    # Turno 1 (Model's Turn): se decodifica SIN quitar tokens especiales,
    # porque necesitamos ver <|tool_call>...<tool_call|> para poder parsearlo.
    raw = _generate(messages, max_new_tokens, skip_special=False)

    parsed = None
    try:
        parsed = processor.parse_response(raw)
        tool_calls = [
            {'name': c['function']['name'], 'arguments': c['function'].get('arguments', {})}
            for c in (parsed.get('tool_calls') or [])
        ]
    except Exception:
        # processor.parse_response no existe todavia en tu version de
        # transformers, o no reconocio el formato: usamos el regex de respaldo.
        tool_calls = _extract_tool_calls_fallback(raw)

    if not tool_calls:
        contenido = (parsed.get('content') if isinstance(parsed, dict) else None)
        if not contenido:
            contenido = processor.decode(
                processor(text=raw, return_tensors='pt')['input_ids'][0],
                skip_special_tokens=True,
            )
        return {'respuesta': contenido.strip(), 'tool_calls': [], 'tool_resultados': []}

    # Turno 2 (Developer's Turn): ejecutamos las tools reales.
    tool_functions = tool_functions or {}
    resultados = []
    for call in tool_calls:
        nombre = call['name']
        args = call.get('arguments', {})
        fn = tool_functions.get(nombre)
        if fn is None:
            resultados.append({
                'name': nombre,
                'response': {'error': f"Tool '{nombre}' no esta registrada o no esta permitida para este perfil."},
            })
            continue
        try:
            resultados.append({'name': nombre, 'response': fn(**args)})
        except Exception as e:
            resultados.append({'name': nombre, 'response': {'error': str(e)}})

    messages.append({
        'role': 'assistant',
        'tool_calls': [{'function': c} for c in tool_calls],
        'tool_responses': resultados,
    })

    # Turno 3 (Final Response): Gemma redacta la respuesta pedagogica final
    # ya con el resultado real de la tool. Aqui si quitamos tokens especiales.
    respuesta_final = _generate(messages, max_new_tokens_final, skip_special=True)
    return {'respuesta': respuesta_final.strip(), 'tool_calls': tool_calls, 'tool_resultados': resultados}


AUDIO_TRANSCRIPTION_SYSTEM = r'''
Eres un transcriptor de audio a texto. Tu UNICA salida debe ser la transcripcion
literal de lo que se dice en el audio, en espanol, lo mas fiel posible palabra por
palabra. No traduzcas, no resumas, no corrijas la gramatica del hablante, no agregues
comentarios ni texto antes o despues de la transcripcion. Si el audio no tiene voz
humana entendible, responde exactamente: [audio_no_entendible]
'''


def gemma_audio_generate(system_prompt: str, audio_path: str, user_text: str = 'Sigue las instrucciones del system prompt sobre este audio.', max_new_tokens: int = 900, temperature: float = 0.3) -> str:
    """Motor real de audio de este backend (reemplaza a Whisper). Gemma 4 acepta
    audio de forma nativa (ademas de texto e imagen), asi que en vez de
    transcribir primero con un modelo de ASR aparte y despues mandarle el
    texto a Gemma, aqui se le manda el audio directo junto con el system
    prompt de la tarea (perfil, transcripcion, experiencia, etc.). Esto ahorra
    una pasada de inferencia y deja que Gemma razone sobre el audio crudo
    (tono, pausas, enfasis) en vez de sobre una transcripcion ya aplanada.

    Usada por: transcribe_audio_safe (transcripcion literal) y
    build_profile_from_audio (perfil directo desde el audio, un solo paso).
    """
    content = [
        {'type': 'audio', 'audio': str(audio_path)},
        {'type': 'text', 'text': user_text},
    ]
    messages = [
        {'role': 'system', 'content': system_prompt},
        {'role': 'user', 'content': content},
    ]
    inputs = processor.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors='pt'
    ).to(model.device, dtype=model.dtype if hasattr(model, 'dtype') else torch.bfloat16)
    input_len = inputs['input_ids'].shape[-1]
    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=temperature,
            top_p=0.9,
        )
    return processor.decode(output_ids[0][input_len:], skip_special_tokens=True).strip()
````

````python
class LearnerProfile(BaseModel):
    nombre: str = 'Estudiante'
    objetivo: str = ''
    nivel_autopercibido: Literal['principiante', 'intermedio', 'avanzado'] = 'principiante'
    estilo_preferido: str = ''
    formato_preferido: str = ''
    duracion_sesion: str = ''
    puntos_de_bloqueo: List[str] = Field(default_factory=list)
    motivacion: str = ''
    tono_preferido: str = 'amigable'
    preferencias_extra: str = ''
    # NUEVO: permite personalizar la experiencia (ej. adulto mayor -> interfaces simples).
    # Todos los campos de arriba ahora son opcionales para que un audio corto o incompleto
    # igual genere un perfil utilizable; lo unico que realmente importa es el nombre.
    grupo_etario: Literal['nino', 'adolescente', 'adulto', 'adulto_mayor'] = 'adulto'
    necesidades_especiales: str = ''


class Subtema(BaseModel):
    nombre: str
    prerrequisitos: List[str] = Field(default_factory=list)
    errores_comunes: List[str] = Field(default_factory=list)
    tipo_de_concepto: Literal['declarativo', 'procedimental', 'dinamico'] = 'declarativo'


class TemaCurricular(BaseModel):
    tema: str
    subtemas: List[Subtema] = Field(default_factory=list)


class Materia(BaseModel):
    materia: str
    unidad: str
    temas: List[TemaCurricular] = Field(default_factory=list)


class EjemploNivel(BaseModel):
    nivel: Literal['principiante', 'intermedio', 'avanzado']
    enunciado: str
    pista: str
    respuesta: str


class MicroReto(BaseModel):
    pregunta: str
    opciones: List[str]
    respuesta_correcta: int
    explicacion: str


class Leccion(BaseModel):
    titulo: str
    tema_padre: str
    subtema: str
    nivel_tecnico: Literal['principiante', 'intermedio', 'avanzado'] = 'principiante'
    enfoque_usuario: str = ''
    conceptos_previos: List[str] = Field(default_factory=list)
    mapa_conocimiento: List[str] = Field(default_factory=list)
    explicacion_paso_a_paso: List[str] = Field(default_factory=list)
    pausa_reflexion: str = ''
    glosario: Dict[str, str] = Field(default_factory=dict)
    micro_retos: List[MicroReto] = Field(default_factory=list)
    ejemplos: List[EjemploNivel] = Field(default_factory=list)
    historia_base: str = ''
    guion_audio: str = ''
    latex_clave: List[str] = Field(default_factory=list)
    prompt_ilustracion: str = ''
    html_prompt: str = ''
    recomendacion_visual: Literal['imagen', 'html', 'combinada'] = 'html'
    visual_tipo: Literal['integral', 'complex', 'function', 'geometry', 'default'] = 'default'
    liked: bool = False
    version: int = 1
    feedback_historial: List[str] = Field(default_factory=list)


@dataclass
class PageBlock:
    page: int
    text: str
    image_paths: List[str] = field(default_factory=list)
````

## 6. Perfil del estudiante y temario desde PDF

````python
_NOMBRE_PATTERNS = [
    r"\bme llamo\s+([A-ZÁÉÍÓÚÑ][a-zA-ZÁÉÍÓÚÑñ]+)",
    r"\bmi nombre es\s+([A-ZÁÉÍÓÚÑ][a-zA-ZÁÉÍÓÚÑñ]+)",
    r"\bllamame\s+([A-ZÁÉÍÓÚÑ][a-zA-ZÁÉÍÓÚÑñ]+)",
    r"\bsoy\s+([A-ZÁÉÍÓÚÑ][a-zA-ZÁÉÍÓÚÑñ]+)\b(?!\s+(?:un|una|el|la|de|estudiante|nuevo|nueva))",
]


def _regex_fallback_nombre(texto: str) -> str:
    """Red de seguridad, no un reemplazo de Gemma: solo se usa si Gemma dejo
    'nombre' vacio en su JSON a pesar de que el texto si menciona un nombre de
    forma explicita. Si Gemma ya devolvio algo, esta funcion nunca se llama."""
    for patron in _NOMBRE_PATTERNS:
        m = re.search(patron, texto, flags=re.IGNORECASE)
        if m:
            return m.group(1).strip().capitalize()
    return ''


_NIVELES_VALIDOS = {'principiante', 'intermedio', 'avanzado'}
_GRUPOS_VALIDOS = {'nino', 'adolescente', 'adulto', 'adulto_mayor'}


def _sanitize_profile_data(data: dict, debug_label: str = '') -> dict:
    """Red de seguridad REAL contra el bug de 'campo Literal vacio'.

    PROFILE_SYSTEM le pide a Gemma dejar '' en los campos donde no hay
    informacion, pero 'nivel_autopercibido' y 'grupo_etario' son Literal en
    LearnerProfile (Pydantic), no texto libre: una cadena vacia ahi NO es un
    valor valido y Pydantic revienta con un 400 antes de siquiera crear el
    perfil (esto es exactamente lo que pasaba en /profile/audio con audios
    cortos o ambiguos, donde Gemma no tenia certeza y devolvia '').

    Esta funcion corrige eso ANTES de construir LearnerProfile: si el valor
    que devolvio Gemma no es exactamente una de las opciones permitidas
    (vacio, mal escrito, en mayusculas, sinonimo no contemplado, etc.), se
    reemplaza por el default mas conservador en vez de dejar que Pydantic
    truene el endpoint completo.
    """
    nivel = str(data.get('nivel_autopercibido') or '').strip().lower()
    if nivel not in _NIVELES_VALIDOS:
        if nivel:
            print(f"[_sanitize_profile_data:{debug_label}] 'nivel_autopercibido' invalido ('{nivel}'), uso 'principiante' por defecto.")
        data['nivel_autopercibido'] = 'principiante'
    else:
        data['nivel_autopercibido'] = nivel

    grupo = str(data.get('grupo_etario') or '').strip().lower()
    if grupo not in _GRUPOS_VALIDOS:
        if grupo:
            print(f"[_sanitize_profile_data:{debug_label}] 'grupo_etario' invalido ('{grupo}'), uso 'adulto' por defecto.")
        data['grupo_etario'] = 'adulto'
    else:
        data['grupo_etario'] = grupo

    # Robustez extra: si Gemma devuelve 'puntos_de_bloqueo' como string en vez
    # de lista (tambien pasa de vez en cuando con audios cortos), lo envolvemos
    # en vez de dejar que Pydantic truene por tipo incorrecto.
    bloqueos = data.get('puntos_de_bloqueo')
    if isinstance(bloqueos, str):
        data['puntos_de_bloqueo'] = [bloqueos] if bloqueos.strip() else []
    elif bloqueos is None:
        data['puntos_de_bloqueo'] = []

    return data


def build_profile_from_interview(interview_text: str) -> LearnerProfile:
    interview_text = (interview_text or '').strip()
    if not interview_text:
        raise ValueError('No hay texto de entrevista para construir el perfil.')

    # temperature baja: es extraccion estructurada (seguir las reglas deterministas
    # de PROFILE_SYSTEM), no generacion creativa.
    raw = gemma_generate(
        PROFILE_SYSTEM,
        f'Entrevista transcrita del estudiante:\n{interview_text}',
        max_new_tokens=1800,
        temperature=0.15,
    )
    # required_keys=['nombre'] hace que quede loggeado en consola cuando Gemma
    # no lo llena, en vez de fallar en silencio via el default de Pydantic.
    data = extract_json_block(raw, required_keys=['nombre'], debug_label='perfil')

    if not str(data.get('nombre') or '').strip():
        rescatado = _regex_fallback_nombre(interview_text)
        data['nombre'] = rescatado or 'Estudiante'
        print(f"[build_profile_from_interview] Gemma no devolvio 'nombre'; se uso fallback regex -> '{data['nombre']}'")

    data = _sanitize_profile_data(data, debug_label='perfil_texto')
    profile = LearnerProfile(**data)
    write_json(PROFILE_PATH, profile.model_dump())
    return profile


def build_profile_from_audio(audio_path: str, contexto_extra: str = '') -> LearnerProfile:
    audio_path = str(audio_path or '').strip()
    if not audio_path:
        raise ValueError('No hay audio para construir el perfil.')

    transcript = transcribe_audio_safe(audio_path)
    instruccion = f'Este es el texto transcrito de la entrevista del estudiante: {transcript}. Genera el perfil segun las instrucciones del system prompt.'
    if contexto_extra and contexto_extra.strip():
        instruccion += f'Contexto adicional en texto: {contexto_extra.strip()}'

    raw = gemma_generate(
        PROFILE_SYSTEM,
        instruccion,
        max_new_tokens=1800,
    )
    data = extract_json_block(raw, required_keys=['nombre'], debug_label='perfil_audio')

    if not str(data.get('nombre') or '').strip():
        rescatado = _regex_fallback_nombre(transcript)
        data['nombre'] = rescatado or 'Estudiante'
        print("[build_profile_from_audio] Gemma no devolvio 'nombre'; se uso fallback regex o 'Estudiante'.")

    profile = LearnerProfile(**data)
    write_json(PROFILE_PATH, profile.model_dump())
    return profile


def load_profile() -> Optional[LearnerProfile]:
    if not PROFILE_PATH.exists():
        return None
    try:
        return LearnerProfile(**read_json(PROFILE_PATH, {}))
    except Exception:
        return None


def extract_pdf_content(pdf_path: str) -> List[PageBlock]:
    doc = fitz.open(pdf_path)
    blocks = []
    for page_idx, page in enumerate(doc):
        text = page.get_text('text')
        pix = page.get_pixmap(matrix=fitz.Matrix(1.4, 1.4))
        img_path = ASSETS / f'page_{page_idx + 1}.png'
        pix.save(str(img_path))
        blocks.append(PageBlock(page=page_idx + 1, text=text, image_paths=[str(img_path)]))
    doc.close()
    return blocks


def pdf_blocks_to_text(blocks: List[PageBlock], max_chars: int = 18000) -> str:
    return '\n\n'.join([f'PAGINA {b.page}\n{b.text}' for b in blocks])[:max_chars]


def build_curriculum_from_pdf_text(pdf_text: str) -> Materia:
    raw = gemma_generate(CURRICULUM_SYSTEM, f'Texto del temario:\n{pdf_text}', max_new_tokens=2600)
    return Materia(**extract_json_block(raw))
````

## 7. Progreso y favoritos

````python
def mark_progress(subtopic: str, status: str):
    progress = read_json(PROGRESS_PATH, {})
    progress[subtopic] = {
        'status': status,
        'updated_at': datetime.datetime.now().isoformat(timespec='seconds')
    }
    write_json(PROGRESS_PATH, progress)


def get_progress() -> dict:
    return read_json(PROGRESS_PATH, {})


def save_favorite(title: str, kind: str, path: str, subtopic: str, reason: str = 'me gusto'):
    favs = read_json(FAVORITES_PATH, [])
    favs.append({
        'title': title,
        'kind': kind,
        'path': path,
        'subtopic': subtopic,
        'reason': reason,
        'timestamp': datetime.datetime.now().isoformat(timespec='seconds')
    })
    write_json(FAVORITES_PATH, favs)


def get_favorites() -> list:
    return read_json(FAVORITES_PATH, [])
````

````python
def generate_lesson(profile: LearnerProfile, tema_padre: str, subtema: Subtema, nivel: str = 'principiante', enfoque: str = '') -> Leccion:
    progress = read_json(PROGRESS_PATH, {})
    prompt = f'''Perfil del estudiante:
{json.dumps(profile.model_dump(), ensure_ascii=False, indent=2)}

Tema padre: {tema_padre}
Subtema: {subtema.nombre}
Tipo de concepto: {subtema.tipo_de_concepto}
Prerrequisitos: {", ".join(subtema.prerrequisitos) or "ninguno"}
Errores comunes: {", ".join(subtema.errores_comunes) or "ninguno"}
Nivel tecnico solicitado: {nivel}
Enfoque adicional: {enfoque or "sin enfoque adicional"}
Estado de progreso actual: {json.dumps(progress.get(subtema.nombre, {}), ensure_ascii=False)}'''
    raw = gemma_generate(LESSON_SYSTEM, prompt, max_new_tokens=3600)
    data = extract_json_block(raw)
    lesson = Leccion(
        titulo=data['titulo'],
        tema_padre=tema_padre,
        subtema=subtema.nombre,
        nivel_tecnico=nivel,
        enfoque_usuario=enfoque,
        conceptos_previos=data.get('conceptos_previos', []),
        mapa_conocimiento=data.get('mapa_conocimiento', []),
        explicacion_paso_a_paso=data.get('explicacion_paso_a_paso', []),
        pausa_reflexion=data.get('pausa_reflexion', ''),
        glosario=data.get('glosario', {}),
        micro_retos=[MicroReto(**r) for r in data.get('micro_retos', [])],
        ejemplos=[EjemploNivel(**e) for e in data.get('ejemplos', [])],
        historia_base=data.get('historia_base', ''),
        guion_audio=data.get('guion_audio', ''),
        latex_clave=data.get('latex_clave', []),
        prompt_ilustracion=data.get('prompt_ilustracion', ''),
        html_prompt=data.get('html_prompt', ''),
        recomendacion_visual=data.get('recomendacion_visual', 'html'),
        visual_tipo=data.get('visual_tipo', 'default'),
    )
    memory = read_json(LESSONS_PATH, {})
    memory[subtema.nombre] = lesson.model_dump()
    write_json(LESSONS_PATH, memory)
    mark_progress(subtema.nombre, 'started')
    return lesson


def apply_feedback(lesson: Leccion, comment: str) -> Leccion:
    payload = {'lesson': lesson.model_dump(), 'feedback': comment}
    raw = gemma_generate(FEEDBACK_SYSTEM, json.dumps(payload, ensure_ascii=False), max_new_tokens=3200)
    data = extract_json_block(raw)
    updated = Leccion(**{**lesson.model_dump(), **data})
    updated.version = lesson.version + 1
    updated.feedback_historial = lesson.feedback_historial + [comment]
    memory = read_json(LESSONS_PATH, {})
    memory[lesson.subtema] = updated.model_dump()
    write_json(LESSONS_PATH, memory)
    return updated


def load_lesson(subtema: str) -> Optional[Leccion]:
    memory = read_json(LESSONS_PATH, {})
    data = memory.get(subtema)
    return Leccion(**data) if data else None
````

## 9. Lectura de apuntes por imagen (multimodal)

````python
def extract_notes_from_images(image_paths: List[str]) -> dict:
    default = {
        'respuesta_natural': '',
        'resumen': '',
        'conceptos': [],
        'dudas': [],
        'texto_fuente': '',
    }
    if not image_paths:
        return default

    user_prompt = (
        'Lee estas fotos de apuntes y explicale al estudiante, en tono natural y hablado, que encontraste. '
        'Si hay formulas, terminos o fragmentos incompletos, interpretalos con cautela.'
    )
    raw = gemma_multimodal_generate(NOTES_SYSTEM, user_prompt, image_paths, max_new_tokens=2200)

    try:
        data = extract_json_block(raw, required_keys=['respuesta_natural', 'texto_fuente'], debug_label='notas_imagen')
    except ValueError:
        # Gemma no devolvio JSON valido: no tiramos el trabajo del modelo, usamos su
        # texto crudo como respuesta natural para que el usuario reciba algo util
        # de todas formas, y lo dejamos loggeado para poder depurar el prompt.
        print('[extract_notes_from_images] Gemma no devolvio JSON valido, usando texto crudo como respaldo.')
        texto_crudo = raw.strip()
        data = {'respuesta_natural': texto_crudo[:800], 'resumen': texto_crudo[:400]}

    merged = {**default, **data}
    write_json(NOTE_IMAGES_PATH, {'images': image_paths, 'analysis': merged})
    return merged


def enrich_lesson_with_notes(lesson: Leccion, notes_data: dict) -> Leccion:
    if not notes_data or not notes_data.get('texto_fuente'):
        return lesson
    payload = {
        'lesson': lesson.model_dump(),
        'notes': notes_data,
        'instruction': 'Reescribe y enriquece la leccion usando los apuntes del estudiante como contexto prioritario. Manten estructura didactica y micro-retos.'
    }
    raw = gemma_generate(FEEDBACK_SYSTEM, json.dumps(payload, ensure_ascii=False), max_new_tokens=3200)
    data = extract_json_block(raw)
    updated = Leccion(**{**lesson.model_dump(), **data})
    updated.version = lesson.version + 1
    updated.feedback_historial = lesson.feedback_historial + ['Enriquecida con apuntes por imagen']
    memory = read_json(LESSONS_PATH, {})
    memory[lesson.subtema] = updated.model_dump()
    write_json(LESSONS_PATH, memory)
    return updated
````

## 11 bis. Generacion de experiencias personalizadas (Gemma escribe su propio HTML)

Nueva capacidad: a partir de una instruccion en texto o en audio (ej. "escribeme un cuento sobre esto", "ayudame a leer este tema"), Gemma 4 decide el tipo de experiencia, redacta el contenido usando los apuntes/temario ya cargados como fuente, y genera su propia pagina HTML autocontenida. Si el perfil activo es de un adulto mayor (o pide algo simple), Gemma recibe instrucciones explicitas para producir una interfaz muy sencilla, con letras grandes y minima navegacion.

No se usa JSON para este contenido (el HTML generado rompe facilmente el formato JSON), sino un formato de bloques delimitados que se parsea con una funcion propia, con un HTML de respaldo por si Gemma no sigue el formato al pie de la letra.


````python
EXPERIENCES_PATH = DATA / 'experiences.json'
if not EXPERIENCES_PATH.exists():
    write_json(EXPERIENCES_PATH, [])


def _collect_apuntes_context(subtema: Optional[str] = None, max_chars: int = 6000) -> str:
    partes = []
    materia = STATE.get('materia')
    if materia:
        partes.append(f'Materia: {materia.materia} - Unidad: {materia.unidad}')
        for tema in materia.temas:
            nombres = ', '.join([s.nombre for s in tema.subtemas])
            partes.append(f'Tema: {tema.tema} -> Subtemas: {nombres}')
    notas = read_json(NOTE_IMAGES_PATH, {})
    analysis = notas.get('analysis') if isinstance(notas, dict) else None
    if analysis:
        if analysis.get('resumen'):
            partes.append(f"Resumen de apuntes en imagen: {analysis['resumen']}")
        if analysis.get('texto_fuente'):
            partes.append(f"Texto de apuntes: {analysis['texto_fuente']}")
    if subtema:
        lesson = load_lesson(subtema)
        if lesson:
            partes.append(f'Leccion existente sobre "{subtema}": ' + '\n'.join(lesson.explicacion_paso_a_paso))
    texto = '\n\n'.join(partes)
    return texto[:max_chars] if texto else 'No hay apuntes ni temario cargados todavia.'


def _parse_experience_blocks(raw: str) -> dict:
    """Parsea el formato de bloques ===MARCADOR=== que escribe Gemma.

    OPTIMIZACION: ya no hay bloque ===HTML=== que parsear (Gemma dejo de generar
    HTML, ver EXPERIENCE_SYSTEM). Esto simplifica el parseo y elimina el bug
    original de HTML cortado, porque ya no hay HTML que se pueda cortar.
    """
    raw = raw.strip()

    def _grab(marker_ini, marker_fin, text):
        start = text.find(marker_ini)
        if start == -1:
            return ''
        start += len(marker_ini)
        end = text.find(marker_fin, start)
        chunk = text[start:end] if end != -1 else text[start:]
        return chunk.strip()

    tipo = _grab('===TIPO===', '===TITULO===', raw) or 'experiencia'
    titulo = _grab('===TITULO===', '===TEXTO===', raw) or 'Experiencia personalizada'
    texto = _grab('===TEXTO===', '===AUDIO===', raw)
    audio_guion = _grab('===AUDIO===', '===FIN===', raw)

    if not texto:
        # Respaldo: si Gemma no siguio el formato de marcadores al pie de la
        # letra, igual mostramos su salida cruda en vez de dejar la experiencia vacia.
        print('[_parse_experience_blocks] No se encontro ===TEXTO===, usando salida cruda como respaldo.')
        texto = raw

    if audio_guion.strip().upper() == 'NINGUNO':
        audio_guion = ''

    return {
        'tipo_experiencia': tipo.strip(),
        'titulo': titulo.strip(),
        'contenido_texto': texto.strip(),
        'guion_audio': audio_guion.strip(),
    }


def _render_experience_html(tipo: str, titulo: str, texto: str, profile: LearnerProfile) -> str:
    """Arma el HTML en Python (instantaneo, sin gastar tokens de Gemma en escribir
    markup) respetando el perfil: letras grandes y diseno simple para adulto_mayor
    o necesidades_especiales, tono calido para nino, diseno limpio en el resto.
    Antes esta logica se le pedia a Gemma en texto libre; ahora es codigo, asi
    que es consistente en cada llamada y no consume tokens de generacion."""
    parrafos = [p.strip() for p in texto.split('\n\n') if p.strip()]
    cuerpo = '\n'.join(f'<p>{p}</p>' for p in parrafos) or f'<p>{texto}</p>'

    simple = profile.grupo_etario == 'adulto_mayor' or bool((profile.necesidades_especiales or '').strip())
    ninos = profile.grupo_etario == 'nino'

    if simple:
        font_size, titulo_size, bg, accent, max_w = '24px', '38px', '#ffffff', '#1d4ed8', '760px'
    elif ninos:
        font_size, titulo_size, bg, accent, max_w = '19px', '30px', '#fff7ed', '#ea580c', '820px'
    else:
        font_size, titulo_size, bg, accent, max_w = '17px', '28px', '#f5f7fb', '#2563eb', '900px'

    return f'''<!DOCTYPE html>
<html lang="es"><head><meta charset="utf-8">
<meta name="viewport" content="width=device-width, initial-scale=1" />
<title>{titulo}</title>
<style>
body{{font-family:Arial,sans-serif;background:{bg};color:#17212b;padding:32px;line-height:1.7;font-size:{font_size};}}
.wrap{{max-width:{max_w};margin:0 auto;background:#fff;padding:32px;border-radius:16px;box-shadow:0 2px 12px rgba(0,0,0,0.06);}}
h1{{font-size:{titulo_size};color:{accent};margin-top:0;}}
.tag{{display:inline-block;background:{accent};color:#fff;padding:4px 12px;border-radius:999px;font-size:14px;margin-bottom:16px;}}
p{{margin:0 0 18px 0;}}
</style></head><body><div class="wrap">
<span class="tag">{tipo}</span>
<h1>{titulo}</h1>
{cuerpo}
</div></body></html>'''


def generate_experience(profile: LearnerProfile, instruccion: str, contexto_extra: str = '', subtema: Optional[str] = None) -> dict:
    apuntes = _collect_apuntes_context(subtema)

    # PASO NUEVO (tool calling real dentro de /experience): antes esta funcion
    # solo llamaba a gemma_generate (una sola pasada, SIN tools). Si el usuario
    # pedia "busca X" o "analiza esta pagina", Gemma no tenia forma de hacerlo
    # de verdad: solo redactaba texto que SONABA a una busqueda, inventado.
    # Ahora se hace una pasada previa con gemma_chat_with_tools (misma funcion
    # que usa /assistant), con las tools reales del perfil (calculadora,
    # buscar_en_internet, leer_pagina_web segun grupo_etario). Si Gemma decide
    # usar una tool aqui, el resultado REAL se agrega como contexto adicional
    # antes de redactar la experiencia final.
    tool_calls, tool_resultados, contexto_de_tools = _gather_tool_context_for_experience(
        profile, instruccion, contexto_extra
    )

    prompt = f'''Perfil del estudiante:
{json.dumps(profile.model_dump(), ensure_ascii=False, indent=2)}

Instruccion del estudiante (que experiencia quiere ahora mismo):
{instruccion}

Contexto adicional dado por el estudiante:
{contexto_extra or "sin contexto adicional"}

Informacion real obtenida con herramientas (busqueda web / lectura de pagina / calculadora,
si se uso alguna; usa esto como fuente de verdad, no lo inventes de nuevo):
{contexto_de_tools or "no se uso ninguna herramienta para esta experiencia"}

Apuntes / temario disponible como fuente de contenido:
{apuntes}'''

    # OPTIMIZACION CLAVE: antes max_new_tokens=6000 porque Gemma tambien tenia que
    # escribir un HTML completo autocontenido (facilmente >4000 tokens el HTML solo).
    # Ahora Gemma SOLO redacta contenido (el HTML lo arma _render_experience_html en
    # Python), asi que 1400 tokens sobran de sobra y la generacion es varias veces
    # mas rapida y ya no se corta a la mitad.
    raw = gemma_generate(EXPERIENCE_SYSTEM, prompt, max_new_tokens=1400, temperature=0.45)
    parsed = _parse_experience_blocks(raw)
    if not parsed['contenido_texto']:
        print('[generate_experience] Advertencia: Gemma no genero contenido_texto para esta experiencia.')

    html = _render_experience_html(parsed['tipo_experiencia'], parsed['titulo'], parsed['contenido_texto'], profile)

    slug = _slug(parsed['titulo'] or instruccion)
    html_path = HTML / f'experiencia_{slug}_{uuid.uuid4().hex[:8]}.html'
    html_path.write_text(html, encoding='utf-8')

    audio_path = None
    if parsed['guion_audio']:
        audio_path = text_to_audio_path(parsed['guion_audio'], name_hint=f'experiencia_{slug}')

    registro = {
        'id': uuid.uuid4().hex[:10],
        'tipo_experiencia': parsed['tipo_experiencia'],
        'titulo': parsed['titulo'],
        'instruccion': instruccion,
        'subtema': subtema or '',
        'contenido_texto': parsed['contenido_texto'],
        'html_path': str(html_path),
        'audio_path': audio_path,
        'grupo_etario_usado': profile.grupo_etario,
        'tool_calls': tool_calls,
        'tool_resultados': tool_resultados,
        'timestamp': datetime.datetime.now().isoformat(timespec='seconds'),
    }
    historial = read_json(EXPERIENCES_PATH, [])
    historial.append(registro)
    write_json(EXPERIENCES_PATH, historial)
    return registro
````

````python
from fastapi import FastAPI, UploadFile, File, Form, HTTPException
from fastapi.middleware.cors import CORSMiddleware
from fastapi.responses import FileResponse, JSONResponse
from fastapi.staticfiles import StaticFiles
from pydantic import BaseModel as _BM
from typing import List as _List, Optional as _Opt

app = FastAPI(title='Gemma 4 Study Companion API', version='1.0')

app.add_middleware(
    CORSMiddleware,
    allow_origins=['*'],
    allow_credentials=True,
    allow_methods=['*'],
    allow_headers=['*'],
)

# Sirve los archivos generados directamente por URL (audio, html, imagenes)
app.mount('/files/media', StaticFiles(directory=str(MEDIA)), name='media')
app.mount('/files/html', StaticFiles(directory=str(HTML)), name='html')
app.mount('/files/assets', StaticFiles(directory=str(ASSETS)), name='assets')
app.mount('/files/exports', StaticFiles(directory=str(EXPORT_DIR)), name='exports')


def _save_upload(upload: UploadFile, folder: Path) -> str:
    ext = Path(upload.filename or '').suffix or ''
    dest = folder / f'{uuid.uuid4().hex[:10]}{ext}'
    with open(dest, 'wb') as f:
        f.write(upload.file.read())
    return str(dest)


# ---------- Esquemas de request ----------
class InterviewBody(_BM):
    texto: str


class TopicBody(_BM):
    topic_index: int
    nivel: Literal['principiante', 'intermedio', 'avanzado'] = 'principiante'
    enfoque: str = ''


class FeedbackBody(_BM):
    comentario: str


class QuizBody(_BM):
    respuesta_index: int


class ProgressBody(_BM):
    subtema: str
    estado: str


class FavoriteBody(_BM):
    subtema: str
    motivo: str = 'me gusto'


class ExportBody(_BM):
    subtema: str
    formato: Literal['txt', 'latex'] = 'txt'
    contenido: _Opt[str] = None


class AskBody(_BM):
    pregunta: str
    contexto: str = ''


# ---------- Salud ----------
@app.get('/health')
def health():
    return {'status': 'ok', 'model': MODEL_ID, 'gpu': torch.cuda.is_available()}


# ---------- Perfil ----------
@app.post('/profile/interview')
def profile_interview(body: InterviewBody):
    try:
        profile = build_profile_from_interview(body.texto)
        STATE['profile'] = profile
        return profile.model_dump()
    except Exception as e:
        raise HTTPException(400, f'No se pudo construir el perfil: {e}')


@app.post('/profile/audio')
async def profile_audio(
    audio: UploadFile = File(...),
    texto_extra: str = Form(''),
    incluir_transcripcion: bool = Form(False),
):
    """CAMBIO: ya no transcribe con Whisper y despues manda texto a Gemma (dos
    pasos, dos modelos). Ahora manda el audio directo a build_profile_from_audio,
    que usa Gemma 4 en un solo paso (audio -> JSON de perfil).

    incluir_transcripcion=true agrega una llamada EXTRA a Gemma solo para
    devolver la transcripcion literal en la respuesta (util si tu frontend
    quiere mostrar 'esto fue lo que escuchamos'). Por defecto viene apagado
    para mantener la respuesta lo mas rapida posible y no arriesgar el tunel
    de ngrok con una segunda inferencia de audio innecesaria.
    """
    path = _save_upload(audio, UPLOADS)
    try:
        profile = build_profile_from_audio(path, contexto_extra=texto_extra)
        STATE['profile'] = profile
        resp = {'perfil': profile.model_dump()}
        if incluir_transcripcion:
            resp['transcripcion'] = transcribe_audio_safe(path)
        return resp
    except Exception as e:
        raise HTTPException(400, f'No se pudo construir el perfil: {e}')


@app.get('/profile')
def get_profile():
    profile = STATE.get('profile') or load_profile()
    if not profile:
        raise HTTPException(404, 'Aun no hay perfil creado.')
    return profile.model_dump()


@app.get('/profile/preguntas')
def get_interview_questions():
    return {'preguntas': INTERVIEW_QUESTIONS}


# ---------- Temario / PDF ----------
@app.post('/curriculum/pdf')
async def curriculum_pdf(pdf: UploadFile = File(...)):
    path = _save_upload(pdf, UPLOADS)
    try:
        blocks = extract_pdf_content(path)
        text = pdf_blocks_to_text(blocks)
        materia = build_curriculum_from_pdf_text(text)
    except Exception as e:
        raise HTTPException(400, f'No se pudo procesar el PDF: {e}')
    flat = []
    for tema in materia.temas:
        for sub in tema.subtemas:
            flat.append((tema.tema, sub))
    STATE['materia'] = materia
    STATE['flat_topics'] = flat
    topics = [{'index': i, 'tema': t, 'subtema': s.nombre} for i, (t, s) in enumerate(flat)]
    return {'materia': materia.model_dump(), 'topics': topics}


@app.get('/curriculum/topics')
def curriculum_topics():
    flat = STATE.get('flat_topics', [])
    return {'topics': [{'index': i, 'tema': t, 'subtema': s.nombre} for i, (t, s) in enumerate(flat)]}


# ---------- Lecciones ----------
@app.post('/lesson')
def create_lesson(body: TopicBody):
    profile = STATE.get('profile') or load_profile()
    if not profile:
        raise HTTPException(400, 'Primero crea el perfil del estudiante en /profile/interview o /profile/audio.')
    try:
        tema, sub = get_flat_topic(body.topic_index)
    except IndexError as e:
        raise HTTPException(400, str(e))
    try:
        lesson = generate_lesson(profile, tema, sub, nivel=body.nivel, enfoque=body.enfoque)
        return lesson.model_dump()
    except Exception as e:
        raise HTTPException(500, f'Fallo generando la leccion: {e}')


@app.get('/lesson/{subtema}')
def get_lesson(subtema: str):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'No existe una leccion generada para ese subtema.')
    return lesson.model_dump()


@app.post('/lesson/{subtema}/feedback')
def lesson_feedback(subtema: str, body: FeedbackBody):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion.')
    try:
        updated = apply_feedback(lesson, body.comentario)
        return updated.model_dump()
    except Exception as e:
        raise HTTPException(500, f'Fallo aplicando feedback: {e}')


@app.post('/lesson/{subtema}/enrich-with-notes')
def lesson_enrich(subtema: str):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion.')
    note_data = read_json(NOTE_IMAGES_PATH, {})
    analysis = note_data.get('analysis') if isinstance(note_data, dict) else None
    if not analysis:
        raise HTTPException(400, 'No hay apuntes cargados. Usa /notes/images primero.')
    updated = enrich_lesson_with_notes(lesson, analysis)
    return updated.model_dump()


@app.get('/lesson/{subtema}/audio')
def lesson_audio(subtema: str):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion.')
    path = generate_audio(lesson)
    return FileResponse(path, media_type='audio/mpeg', filename=Path(path).name)


@app.get('/lesson/{subtema}/html')
def lesson_html(subtema: str):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion.')
    path = generate_practice_html(lesson)
    return FileResponse(path, media_type='text/html', filename=Path(path).name)


@app.get('/lesson/{subtema}/visual')
def lesson_visual(subtema: str):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion.')
    path = build_concept_visual(lesson)
    return FileResponse(path, media_type='text/html', filename=Path(path).name)


@app.get('/lesson/{subtema}/preview-image')
def lesson_preview_image(subtema: str):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion.')
    path = generate_preview_image(lesson)
    return FileResponse(path, media_type='image/png', filename=Path(path).name)


@app.post('/lesson/{subtema}/quiz/verify')
def quiz_verify(subtema: str, body: QuizBody):
    lesson = load_lesson(subtema)
    if not lesson or not lesson.micro_retos:
        raise HTTPException(404, 'No hay micro-retos disponibles para ese subtema.')
    quiz = lesson.micro_retos[-1]
    correcto = body.respuesta_index == quiz.respuesta_correcta
    return {'correcto': correcto, 'explicacion': quiz.explicacion}


@app.post('/lesson/{subtema}/repaso')
def lesson_repaso(subtema: str):
    lesson = load_lesson(subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion.')
    prompt = f'Genera preguntas breves de repaso en espanol para el subtema "{subtema}", basadas en esta leccion:\n{json.dumps(lesson.model_dump(), ensure_ascii=False)}'
    texto = gemma_generate('Eres un tutor que redacta preguntas de repaso claras y breves.', prompt, max_new_tokens=900)
    return {'preguntas_repaso': texto}


# ---------- Apuntes por imagen ----------
@app.post('/notes/images')
async def notes_images(imagenes: _List[UploadFile] = File(...)):
    paths = [_save_upload(img, UPLOADS) for img in imagenes]
    try:
        analysis = extract_notes_from_images(paths)
    except Exception as e:
        raise HTTPException(500, f'Fallo leyendo las imagenes: {e}')

    # El audio se genera a partir de lo que Gemma realmente redacto para el estudiante
    # (respuesta_natural), no de una plantilla fija tipo 'Se cargo la foto. '.
    texto_hablado = (analysis.get('respuesta_natural') or analysis.get('resumen') or 'Ya revise tus apuntes.').strip()
    audio_path = text_to_audio_path(texto_hablado, name_hint='notas')

    return {
        'analisis': analysis,
        'respuesta_natural': analysis.get('respuesta_natural', ''),
        'audio_confirmacion': audio_path,
    }


# ---------- Progreso ----------
@app.post('/progress')
def set_progress(body: ProgressBody):
    mark_progress(body.subtema, body.estado)
    return {'ok': True}


@app.get('/progress')
def all_progress():
    return get_progress()


# ---------- Favoritos ----------
@app.post('/favorites')
def add_favorite(body: FavoriteBody):
    lesson = load_lesson(body.subtema)
    if not lesson:
        raise HTTPException(404, 'Genera primero la leccion antes de marcarla como favorita.')
    html_path = str(HTML / f'{_slug(lesson.titulo)}_v{lesson.version}.html')
    save_favorite(lesson.titulo, 'lesson', html_path, lesson.subtema, body.motivo)
    mark_progress(lesson.subtema, 'liked')
    return {'ok': True}


@app.get('/favorites')
def list_favorites():
    return {'favoritos': get_favorites()}


# ---------- Exportar TXT / LaTeX ----------
@app.post('/export')
def export_text(body: ExportBody):
    lesson = load_lesson(body.subtema)
    content = body.contenido
    if not content:
        if not lesson:
            raise HTTPException(404, 'No hay leccion ni contenido explicito para exportar.')
        content = '\n\n'.join(lesson.explicacion_paso_a_paso) or lesson.historia_base
    path = export_generated_text(body.subtema, content, body.formato)
    return FileResponse(path, media_type='text/plain', filename=Path(path).name)


# ---------- Utilidades libres ----------
@app.post('/audio/transcribe')
async def audio_transcribe(audio: UploadFile = File(...)):
    path = _save_upload(audio, UPLOADS)
    return {'texto': transcribe_audio_safe(path)}


@app.post('/tts')
def tts(texto: str = Form(...)):
    path = text_to_audio_path(texto, name_hint='tts')
    if not path:
        raise HTTPException(400, 'Texto vacio.')
    return FileResponse(path, media_type='audio/mpeg', filename=Path(path).name)


@app.post('/ask')
def ask(body: AskBody):
    system = 'Eres un tutor personal que responde de forma clara, breve y en espanol.'
    prompt = body.pregunta if not body.contexto else f'Contexto: {body.contexto}\n\nPregunta: {body.pregunta}'
    respuesta = gemma_generate(system, prompt, max_new_tokens=1200)
    return {'respuesta': respuesta}
````

### Nuevos endpoints: perfil mas flexible + experiencias personalizadas

No se toco ningun endpoint existente. Se agregan `/experience`, `/experience/audio`, `/experience/historial` y `/experience/{experience_id}`.


````python
class ExperienceBody(_BM):
    instruccion: str
    contexto_extra: str = ''
    subtema: _Opt[str] = None


def _experience_response(registro: dict, transcripcion: _Opt[str] = None) -> dict:
    resp = {
        'id': registro['id'],
        'tipo_experiencia': registro['tipo_experiencia'],
        'titulo': registro['titulo'],
        'subtema': registro['subtema'],
        'contenido_texto': registro['contenido_texto'],
        'url_html': f"/files/html/{Path(registro['html_path']).name}",
        'url_audio': f"/files/media/{Path(registro['audio_path']).name}" if registro['audio_path'] else None,
        'grupo_etario_usado': registro['grupo_etario_usado'],
        # NUEVO: para que el frontend pueda mostrar "Gemma busco esto" / "Gemma leyo esta
        # pagina" en vez de que quede oculto que se uso una tool real.
        'tool_calls': registro.get('tool_calls', []),
        'tool_resultados': registro.get('tool_resultados', []),
        'timestamp': registro['timestamp'],
    }
    if transcripcion is not None:
        resp['transcripcion'] = transcripcion
    return resp


# OPTIMIZACION CLAVE (arregla el "ngrok se muere"): generar una experiencia toma
# varios segundos de GPU aunque ya se recorto el max_new_tokens. Si el endpoint
# se queda esperando (bloqueado) todo ese tiempo antes de responder, el tunel de
# ngrok / el cliente HTTP puede cortar la conexion por inactividad o timeout, y
# el resultado se pierde aunque el modelo si haya terminado de generar.
#
# Con este patron, el endpoint responde CASI INSTANTANEO con un job_id, la
# generacion corre en un hilo en background, y el cliente pregunta el resultado
# con GET /experience/status/{job_id} cada 2-3 segundos hasta que status='done'.
# Asi ninguna conexion HTTP queda abierta mas de una fraccion de segundo.
JOBS: Dict[str, dict] = {}


def _run_experience_job(job_id: str, profile: LearnerProfile, instruccion: str, contexto_extra: str, subtema: _Opt[str], transcripcion: _Opt[str] = None):
    JOBS[job_id]['status'] = 'processing'
    try:
        registro = generate_experience(profile, instruccion, contexto_extra, subtema)
        JOBS[job_id]['status'] = 'done'
        JOBS[job_id]['result'] = _experience_response(registro, transcripcion=transcripcion)
    except Exception as e:
        JOBS[job_id]['status'] = 'error'
        JOBS[job_id]['error'] = str(e)


@app.post('/experience')
def create_experience(body: ExperienceBody):
    profile = STATE.get('profile') or load_profile()
    if not profile:
        raise HTTPException(400, 'Primero crea el perfil del estudiante en /profile/interview o /profile/audio.')
    job_id = uuid.uuid4().hex[:10]
    JOBS[job_id] = {'status': 'pending', 'result': None, 'error': None}
    threading.Thread(
        target=_run_experience_job,
        args=(job_id, profile, body.instruccion, body.contexto_extra, body.subtema),
        daemon=True,
    ).start()
    return {'job_id': job_id, 'status': 'pending', 'poll_url': f'/experience/status/{job_id}'}


@app.post('/experience/audio')
async def create_experience_audio(
    audio: UploadFile = File(...),
    contexto_extra: str = Form(''),
    subtema: _Opt[str] = Form(None),
):
    profile = STATE.get('profile') or load_profile()
    if not profile:
        raise HTTPException(400, 'Primero crea el perfil del estudiante en /profile/interview o /profile/audio.')
    path = _save_upload(audio, UPLOADS)
    transcript = transcribe_audio_safe(path)
    if not transcript.strip():
        raise HTTPException(400, 'No se pudo transcribir el audio.')
    job_id = uuid.uuid4().hex[:10]
    JOBS[job_id] = {'status': 'pending', 'result': None, 'error': None}
    threading.Thread(
        target=_run_experience_job,
        args=(job_id, profile, transcript, contexto_extra, subtema, transcript),
        daemon=True,
    ).start()
    return {'job_id': job_id, 'status': 'pending', 'poll_url': f'/experience/status/{job_id}', 'transcripcion': transcript}


@app.get('/experience/status/{job_id}')
def experience_status(job_id: str):
    job = JOBS.get(job_id)
    if not job:
        raise HTTPException(404, 'job_id no encontrado (revisa que no se haya reiniciado el servidor).')
    return job


@app.get('/experience/historial')
def experience_historial():
    return {'experiencias': read_json(EXPERIENCES_PATH, [])}


@app.get('/experience/{experience_id}')
def experience_get(experience_id: str):
    historial = read_json(EXPERIENCES_PATH, [])
    for r in historial:
        if r['id'] == experience_id:
            return _experience_response(r)
    raise HTTPException(404, 'No existe una experiencia con ese id.')
````

## 16. Optimizaciones aplicadas en esta version (velocidad + estabilidad de ngrok)

**Problema real detectado:** en `/experience`, Gemma tenia que redactar contenido Y una pagina HTML
completa autocontenida en la misma generacion (`max_new_tokens=6000`). Eso son varios minutos de
generacion token a token, y mientras el request HTTP seguia abierto esperando esa respuesta, el tunel
de ngrok se caia (ninguna conexion aguanta bloqueada tanto tiempo sin datos).

**Cambios:**
1. **Gemma ya no escribe HTML** (celdas 13 y 27). Solo redacta `TIPO / TITULO / TEXTO / AUDIO`; el HTML
   se arma en Python (`_render_experience_html`) respetando el perfil (letras grandes para
   `adulto_mayor`, tono calido para `nino`, etc.). `max_new_tokens` bajo de 6000 a 1400.
2. **`/experience` y `/experience/audio` son asincronos** (celda 31): responden de inmediato con un
   `job_id` y la generacion corre en un hilo en background. El cliente consulta el resultado con
   `GET /experience/status/{job_id}` (status: `pending` -> `processing` -> `done`/`error`) cada 2-3
   segundos. Ninguna conexion HTTP/ngrok queda abierta mas de una fraccion de segundo.
3. **`attn_implementation='sdpa'`** al cargar el modelo (celda 7) y `torch.inference_mode()` en vez de
   `no_grad` (celda 9): generacion algo mas rapida sin cambiar la calidad de las respuestas.
4. **Watchdog de ngrok** (celda 35): revisa cada 30s si el tunel sigue vivo y lo reconecta solo si se cae,
   sin que tengas que volver a correr la celda a mano.

### Filtro de temas para busqueda/lectura web

Heuristica simple por palabras clave en varias categorias de contenido daniño. No es
perfecta (alguien podria evadirla con sinonimos o errores ortograficos deliberados), pero
cubre el caso comun de un usuario que sin querer o con mala intencion pide buscar algo
peligroso. Es la misma verificacion para `buscar_en_internet` (antes de buscar) y para
`leer_pagina_web` (antes de pedirle a Firecrawl que lea la URL, y otra vez sobre el
contenido ya descargado, por si la pagina en si trae ese contenido aunque la URL no lo
sugiriera).


````python
_TEMAS_BLOQUEADOS = {
    'armas_y_violencia': [
        'como hacer una bomba', 'fabricar explosivo', 'fabricar arma', 'comprar arma ilegal',
        'como matar', 'como hacer daño a', 'atentado', 'terrorismo',
    ],
    'drogas_ilegales': [
        'como fabricar droga', 'sintetizar droga', 'donde comprar droga', 'como conseguir cocaina',
        'como conseguir metanfetamina',
    ],
    'contenido_sexual_explicito': [
        'porno', 'pornografia', 'contenido sexual explicito', 'sexo explicito',
    ],
    'autolesion': [
        'como suicidarme', 'como lastimarme', 'metodos de suicidio', 'como cortarme',
    ],
    'hacking_y_malware': [
        'como hackear', 'crear virus informatico', 'crear malware', 'robar contraseñas',
        'exploit para hackear',
    ],
}


def _tema_bloqueado(texto: str) -> bool:
    texto_norm = (texto or '').lower()
    for frases in _TEMAS_BLOQUEADOS.values():
        for frase in frases:
            if frase in texto_norm:
                return True
    return False
````

````python
def buscar_en_internet(consulta: str) -> dict:
    """Busca informacion actual en internet (Firecrawl Search) y devuelve
    un resumen de las mejores fuentes encontradas.

    Usala cuando el usuario pida informacion que no esta en el temario ya
    cargado, ejercicios adicionales, o datos que puedan haber cambiado.
    NO la uses para temas peligrosos o ilegales (armas, drogas, violencia,
    contenido sexual, autolesion, hacking): en esos casos disculpate y
    ofrece ayuda con otra cosa.

    Args:
        consulta: Texto de busqueda, 3-8 palabras, en espanol o ingles.

    Returns:
        Un diccionario con 'ok' (bool) y 'resultados' (lista de
        {titulo, resumen, url}), o 'error'.
    """
    if _tema_bloqueado(consulta):
        return {'ok': False, 'error': 'Esta busqueda no esta permitida. Puedo ayudarte con otro tema.'}
    if not FIRECRAWL_API_KEY:
        return {'ok': False, 'error': 'Firecrawl no esta configurado (falta FIRECRAWL_API_KEY en los secrets).'}
    try:
        resp = requests.post(
            'https://api.firecrawl.dev/v1/search',
            headers={'Authorization': f'Bearer {FIRECRAWL_API_KEY}', 'Content-Type': 'application/json'},
            json={'query': consulta, 'limit': 3},
            timeout=20,
        )
        if resp.status_code >= 300:
            return {'ok': False, 'error': f'Firecrawl respondio {resp.status_code}: {resp.text[:300]}'}
        data = resp.json()
        items = (data.get('data') or [])[:3]
        resultados = [
            {
                'titulo': it.get('title', ''),
                'resumen': (it.get('description') or it.get('markdown') or '')[:300],
                'url': it.get('url', ''),
            }
            for it in items
        ]
        return {'ok': True, 'resultados': resultados}
    except Exception as e:
        return {'ok': False, 'error': str(e)}


def leer_pagina_web(url: str) -> dict:
    """Lee y trae el contenido COMPLETO de una pagina web especifica (via
    Firecrawl Scrape), para que puedas explicarselo al usuario paso a paso:

    IMPORTANTE despues de llamar esta tool: tu respuesta final NUNCA debe ser el contenido pegado tal cual... Redacta una explicacion propia, en espanol hablado, organizada.
    por ejemplo instrucciones de un tramite, una guia, un formulario, o
    cualquier pagina que el usuario ya tenga identificada por su URL.

    Usala cuando el usuario te de una URL o te pida "ayudame a entender
    esta pagina" / "como hago este tramite en este sitio" / "leeme que
    dice aqui". No uses esta tool para temas peligrosos o ilegales.

    Args:
        url: Direccion completa de la pagina, ej. "https://www.gob.mx/tramite".

    Returns:
        Un diccionario con 'ok' (bool) y 'titulo'/'contenido' (texto ya
        recortado y listo para resumir), o 'error'.
    """
    if _tema_bloqueado(url):
        return {'ok': False, 'error': 'No puedo leer esa pagina. Puedo ayudarte con otro sitio o tema.'}
    if not FIRECRAWL_API_KEY:
        return {'ok': False, 'error': 'Firecrawl no esta configurado (falta FIRECRAWL_API_KEY en los secrets).'}
    try:
        resp = requests.post(
            'https://api.firecrawl.dev/v1/scrape',
            headers={'Authorization': f'Bearer {FIRECRAWL_API_KEY}', 'Content-Type': 'application/json'},
            json={'url': url, 'formats': ['markdown']},
            timeout=30,
        )
        if resp.status_code >= 300:
            return {'ok': False, 'error': f'Firecrawl respondio {resp.status_code}: {resp.text[:300]}'}
        data = resp.json()
        payload = data.get('data', data)
        contenido = (payload.get('markdown') or '')
        titulo = (payload.get('metadata', {}) or {}).get('title', '')

        if _tema_bloqueado(contenido):
            return {'ok': False, 'error': 'Esta pagina contiene informacion que no puedo compartir. Puedo ayudarte con otro sitio o tema.'}

        return {'ok': True, 'titulo': titulo, 'contenido': contenido[:4000]}
    except Exception as e:
        return {'ok': False, 'error': str(e)}
````

````python
def notion_guardar_nota(titulo: str, contenido: str) -> dict:
    """[PENDIENTE DE ASIGNAR] Guarda una nota o resumen en un cuaderno
    digital de Notion. No esta conectada a ningun perfil todavia: falta
    configurar NOTION_API_KEY / NOTION_PARENT_PAGE_ID y agregarla a
    TOOLS_BY_PROFILE cuando se retome.

    Args:
        titulo: Titulo corto de la nota (maximo 8 palabras).
        contenido: Texto completo de la nota en espanol.

    Returns:
        Un diccionario con 'ok' (bool) y 'mensaje'/'url', o 'error'.
    """
    if not NOTION_API_KEY or not NOTION_PARENT_PAGE_ID:
        return {'ok': False, 'error': 'Notion no esta configurado (falta NOTION_API_KEY o NOTION_PARENT_PAGE_ID en los secrets).'}
    try:
        resp = requests.post(
            'https://api.notion.com/v1/pages',
            headers={
                'Authorization': f'Bearer {NOTION_API_KEY}',
                'Notion-Version': '2022-06-28',
                'Content-Type': 'application/json',
            },
            json={
                'parent': {'page_id': NOTION_PARENT_PAGE_ID},
                'properties': {'title': {'title': [{'text': {'content': titulo[:100]}}]}},
                'children': [{
                    'object': 'block',
                    'type': 'paragraph',
                    'paragraph': {'rich_text': [{'type': 'text', 'text': {'content': contenido[:1900]}}]},
                }],
            },
            timeout=15,
        )
        if resp.status_code >= 300:
            return {'ok': False, 'error': f'Notion respondio {resp.status_code}: {resp.text[:300]}'}
        data = resp.json()
        return {'ok': True, 'mensaje': 'Nota guardada en Notion.', 'url': data.get('url', '')}
    except Exception as e:
        return {'ok': False, 'error': str(e)}


def clima_actual(ciudad: str) -> dict:
    """[PENDIENTE DE ASIGNAR] Consulta el clima actual de una ciudad, via
    Open-Meteo (gratis, sin api key). No esta conectada a ningun perfil
    todavia.

    Args:
        ciudad: Nombre de la ciudad, ej. "Ciudad de Mexico".

    Returns:
        Un diccionario con 'ok' (bool), 'temperatura_c', 'viento_kmh', o 'error'.
    """
    try:
        geo = requests.get(
            'https://geocoding-api.open-meteo.com/v1/search',
            params={'name': ciudad, 'count': 1, 'language': 'es'},
            timeout=10,
        ).json()
        resultados = geo.get('results') or []
        if not resultados:
            return {'ok': False, 'error': f'No encontre la ciudad "{ciudad}".'}
        lat, lon = resultados[0]['latitude'], resultados[0]['longitude']
        nombre_encontrado = resultados[0].get('name', ciudad)
        clima = requests.get(
            'https://api.open-meteo.com/v1/forecast',
            params={'latitude': lat, 'longitude': lon, 'current_weather': 'true'},
            timeout=10,
        ).json()
        actual = clima.get('current_weather', {})
        return {
            'ok': True,
            'ciudad_encontrada': nombre_encontrado,
            'temperatura_c': actual.get('temperature'),
            'viento_kmh': actual.get('windspeed'),
        }
    except Exception as e:
        return {'ok': False, 'error': str(e)}


def recordatorio_crear(texto: str, fecha_hora: str = '') -> dict:
    """[PENDIENTE DE ASIGNAR] Guarda un recordatorio simple, localmente
    (sin depender de ningun servicio externo). No esta conectada a ningun
    perfil todavia.

    Args:
        texto: Que se debe recordar.
        fecha_hora: Cuando, en texto libre. Puede ir vacio.

    Returns:
        Un diccionario con 'ok' (bool) y 'mensaje'.
    """
    try:
        recordatorios = read_json(REMINDERS_PATH, [])
        recordatorios.append({
            'texto': texto,
            'fecha_hora': fecha_hora,
            'creado_en': datetime.datetime.now().isoformat(timespec='seconds'),
        })
        write_json(REMINDERS_PATH, recordatorios)
        return {'ok': True, 'mensaje': f'Listo, te recordare: "{texto}".'}
    except Exception as e:
        return {'ok': False, 'error': str(e)}


_OPS = {
    ast.Add: operator.add, ast.Sub: operator.sub, ast.Mult: operator.mul,
    ast.Div: operator.truediv, ast.Pow: operator.pow, ast.USub: operator.neg,
    ast.Mod: operator.mod,
}
_FUNCS = {
    'sqrt': math.sqrt, 'sin': math.sin, 'cos': math.cos, 'tan': math.tan,
    'log': math.log, 'log10': math.log10, 'exp': math.exp, 'pi': math.pi, 'e': math.e,
}


def _safe_eval(node):
    if isinstance(node, ast.Expression):
        return _safe_eval(node.body)
    if isinstance(node, ast.Constant):
        if not isinstance(node.value, (int, float)):
            raise ValueError('Solo se permiten numeros.')
        return node.value
    if isinstance(node, ast.BinOp) and type(node.op) in _OPS:
        return _OPS[type(node.op)](_safe_eval(node.left), _safe_eval(node.right))
    if isinstance(node, ast.UnaryOp) and type(node.op) in _OPS:
        return _OPS[type(node.op)](_safe_eval(node.operand))
    if isinstance(node, ast.Call) and isinstance(node.func, ast.Name) and node.func.id in _FUNCS:
        return _FUNCS[node.func.id](*[_safe_eval(a) for a in node.args])
    if isinstance(node, ast.Name) and node.id in _FUNCS:
        return _FUNCS[node.id]
    raise ValueError('Expresion no soportada.')


def calculadora(expresion: str) -> dict:
    """Evalua una expresion matematica y devuelve el resultado numerico
    exacto (suma, resta, multiplicacion, division, potencias, raiz,
    trigonometria, logaritmos).

    Usala SIEMPRE que el usuario pida un calculo, una operacion aritmetica,
    o quiera verificar una cuenta que ya hizo, en vez de calcularlo tu
    mismo: tu puedes cometer errores aritmeticos, esta herramienta no.

    Args:
        expresion: Expresion matematica en notacion estandar, ej.
            "2*(3+5)/4" o "sqrt(16)+log(100)".

    Returns:
        Un diccionario con 'ok' (bool) y 'resultado' (float), o 'error'.
    """
    try:
        tree = ast.parse(expresion, mode='eval')
        return {'ok': True, 'resultado': _safe_eval(tree)}
    except Exception as e:
        return {'ok': False, 'error': f'No pude calcular "{expresion}": {e}'}
````

### Registro de tools por perfil

`calculadora` va en todos los perfiles. `buscar_en_internet` y `leer_pagina_web` van en
todos MENOS `nino`. `notion_guardar_nota`, `clima_actual` y `recordatorio_crear` quedan
sin asignar (ver celda anterior).


````python
TOOLS_BY_PROFILE: Dict[str, List[Any]] = {
    'nino': [calculadora, buscar_en_internet, leer_pagina_web],
    'adulto_mayor': [calculadora, buscar_en_internet, leer_pagina_web],
    'adolescente': [calculadora, buscar_en_internet, leer_pagina_web],
    'adulto': [calculadora, buscar_en_internet, leer_pagina_web],
}


def get_tools_for_profile(profile: LearnerProfile):
    tools = TOOLS_BY_PROFILE.get(profile.grupo_etario, [calculadora])
    tool_functions = {t.__name__: t for t in tools}
    return (tools or None), tool_functions


def _assistant_system_for(profile: LearnerProfile) -> str:
    base = (
        f"Eres el tutor personal de {profile.nombre}. Responde siempre en espanol, "
        f"en un tono {profile.tono_preferido or 'amigable'}. Si usas una herramienta, "
        f"nunca muestres detalles tecnicos del resultado (JSON, codigos de error): "
        f"traducelo a una frase natural y calida. Si usaste 'leer_pagina_web': tu respuesta debe ser una EXPLICACION propia del contenido... nunca copies ni pegues el markdown/texto crudo."
    )
    if profile.grupo_etario == 'adulto_mayor':
        base += (
            ' Usa frases cortas y sencillas, evita jerga tecnica, y se muy paciente. '
            'Si el usuario te da una pagina o un tramite, explicaselo paso a paso, '
            'con calma, sin dar por hecho que sabe usar internet. '
            'Si una herramienta falla, nunca muestres el error tecnico: dile con calidez '
            'que no se pudo hacer en este momento pero que no se preocupe.'
        )
    return base
````

### Conectar las tools tambien a `/experience` (no solo a `/assistant`)

Antes, `generate_experience()` (la que arma cuentos/lecturas/experiencias personalizadas)
solo llamaba a `gemma_generate` en una sola pasada, SIN tools. Eso significaba que si el
usuario decia "busca informacion sobre X" o "analiza esta pagina" dentro de una experiencia,
Gemma **no tenia forma real de hacerlo**: redactaba texto que sonaba a resultado de
busqueda, pero inventado (alucinado), no un resultado real de Firecrawl.

`_gather_tool_context_for_experience` hace una pasada previa con `gemma_chat_with_tools`
(la misma funcion de 3 turnos que usa `/assistant`), usando las tools reales del perfil
activo. Si Gemma decide usar `buscar_en_internet`, `leer_pagina_web` o `calculadora` en
esa pasada, el resultado REAL se inyecta como contexto en el prompt de
`generate_experience`, para que la experiencia final se redacte con datos verdaderos y no
inventados.

Costo: esto agrega una generacion extra de GPU a `/experience` (ya era asincrono via
`job_id`, asi que no afecta el timeout de ngrok, solo tarda un poco mas en total).


````python
def _gather_tool_context_for_experience(profile: LearnerProfile, instruccion: str, contexto_extra: str):
    """Pasada previa de tool calling para /experience. Devuelve
    (tool_calls, tool_resultados, contexto_de_tools) para inyectar en el
    prompt de generate_experience.

    Si el perfil no tiene tools asignadas (ej. 'nino'), o Gemma decide que
    no necesita ninguna tool para esta instruccion, devuelve listas vacias
    y un string vacio: generate_experience sigue funcionando exactamente
    igual que antes en ese caso.
    """
    tools, tool_functions = get_tools_for_profile(profile)
    if not tools:
        return [], [], ''

    system = (
        "Eres un asistente que decide si hace falta usar una herramienta "
        "(buscar en internet, leer una pagina web completa, o calcular algo) "
        "para poder preparar el contenido que el usuario pidio. Si la "
        "instruccion menciona una URL, pedir buscar algo, o un calculo, usa "
        "la herramienta correspondiente. Si no hace falta ninguna herramienta "
        "(ej. 'cuentame un cuento sobre fracciones'), responde directamente "
        "sin usar ninguna."
    )
    user_prompt = f"Instruccion del usuario: {instruccion}\nContexto adicional: {contexto_extra or 'ninguno'}"

    resultado = gemma_chat_with_tools(
        system,
        user_prompt,
        tools=tools,
        tool_functions=tool_functions,
        max_new_tokens=500,
        max_new_tokens_final=700,
    )

    tool_calls = resultado.get('tool_calls', [])
    tool_resultados = resultado.get('tool_resultados', [])

    if not tool_calls:
        return [], [], ''

    partes = []
    for r in tool_resultados:
        partes.append(f"Herramienta '{r['name']}' devolvio: {json.dumps(r['response'], ensure_ascii=False)[:1500]}")
    contexto_de_tools = '\n\n'.join(partes)
    return tool_calls, tool_resultados, contexto_de_tools
````

### Endpoint `/assistant` (asincrono, mismo patron de `job_id` que `/experience`)

Igual que en `/experience`, la generacion con tools puede tomar varios segundos (Model's
Turn -> ejecutar tool -> Final Response), asi que el endpoint responde de inmediato con un
`job_id` y el cliente consulta `GET /assistant/status/{job_id}` cada 2-3 segundos. Esto
evita que ngrok corte la conexion por inactividad, igual que ya resolviste para `/experience`.


````python
class AssistantBody(_BM):
    mensaje: str


def _run_assistant_job(job_id: str, profile: LearnerProfile, mensaje: str):
    JOBS[job_id]['status'] = 'processing'
    try:
        tools, tool_functions = get_tools_for_profile(profile)
        resultado = gemma_chat_with_tools(
            _assistant_system_for(profile),
            mensaje,
            tools=tools,
            tool_functions=tool_functions,
            max_new_tokens_final=1300
        )
        JOBS[job_id]['status'] = 'done'
        JOBS[job_id]['result'] = resultado
    except Exception as e:
        JOBS[job_id]['status'] = 'error'
        JOBS[job_id]['error'] = str(e)


@app.post('/assistant')
def assistant(body: AssistantBody):
    """Habla con el tutor usando tool calling nativo de Gemma 4. Las
    herramientas disponibles dependen del grupo_etario del perfil activo
    (ver TOOLS_BY_PROFILE)."""
    profile = STATE.get('profile') or load_profile()
    job_id = uuid.uuid4().hex[:12]
    JOBS[job_id] = {'status': 'pending'}
    threading.Thread(target=_run_assistant_job, args=(job_id, profile, body.mensaje), daemon=True).start()
    return {'job_id': job_id}


@app.get('/assistant/status/{job_id}')
def assistant_status(job_id: str):
    job = JOBS.get(job_id)
    if not job:
        raise HTTPException(status_code=404, detail='job_id no encontrado')
    return job


print("Endpoints registrados: POST /assistant, GET /assistant/status/{job_id}")
print("IMPORTANTE: corre esta celda ANTES de la celda que lanza uvicorn+ngrok, o vuelve a correr esa celda despues.")

@app.post('/assistant/audio')
def assistant_audio(audio: UploadFile = File(...), contexto_extra: str = Form('')):
    """Endpoint para enviar audio al asistente directamente"""
    profile = STATE.get('profile') or load_profile()
    
    # Guardar audio temporalmente
    temp_path = str(UPLOADS / f'temp_{uuid.uuid4().hex[:8]}.webm')
    with open(temp_path, 'wb') as f:
        f.write(audio.file.read())
        
    texto_transcrito = transcribe_audio_safe(temp_path)
    
    job_id = uuid.uuid4().hex[:12]
    JOBS[job_id] = {'status': 'pending'}
    
    # Pasarle el texto transcrito a la funcion normal del asistente
    import threading
    threading.Thread(target=_run_assistant_job, args=(job_id, profile, texto_transcrito), daemon=True).start()
    
    return {'job_id': job_id, 'transcripcion': texto_transcrito}
````
